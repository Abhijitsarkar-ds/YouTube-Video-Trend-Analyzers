Step 2: Load Dataset

```python
import pandas as pd

# Load dataset
df = pd.read_csv(r"c:\Users\abhij\Downloads\archive (8)\INvideos.csv")

# Preview data
print(df.head())
```

Step 3: Data Cleaning

```python
# Remove missing values
df = df.dropna()

# Convert columns if needed
df['views'] = pd.to_numeric(df['views'], errors='coerce')
df['likes'] = pd.to_numeric(df['likes'], errors='coerce')
```

Step 4: Analyze Top Performing Categories

Views by Category

```python


views_category = df.groupby('category_id')['views'].mean().sort_values(ascending=False)
print(views_category)
```

Likes by Category

```python
likes_category = df.groupby('category_id')['likes'].mean().sort_values(ascending=False)
print(likes_category)
```

Step 5: Visualization

Bar Chart (Views)

```python
import matplotlib.pyplot as plt

views_category.plot(kind='bar')
plt.title("Average Views by Category")
plt.xlabel("Category")
plt.ylabel("Views")
plt.xticks(rotation=45)
plt.show()
```

Pie Chart (Top Categories)

```python
views_category.head(5).plot(kind='pie', autopct='%1.1f%%')
plt.title("Top 5 Categories by Views")
plt.ylabel("")
plt.show()
```

Step 6: Engagement Rate

```python
df['engagement'] = (df['likes'] / df['views']) * 100

engagement_category = df.groupby('category_id')['engagement'].mean().sort_values(ascending=False)
print(engagement_category)
```

Step 7: Final Insight Output

```python
print("Top Categories by Views:\n", views_category.head())
print("\nTop Categories by Likes:\n", likes_category.head())
print("\nTop Categories by Engagement:\n", engagement_category.head())
```
